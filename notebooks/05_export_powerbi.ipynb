```python
import sqlite3
from pathlib import Path
import pandas as pd

out = Path("../data/powerbi")
out.mkdir(exist_ok=True)

con = sqlite3.connect("../data/retail.db")
for table in ["customers", "products", "invoices", "invoice_lines"]:
    df = pd.read_sql(f"SELECT * FROM {table}", con)
    df.to_csv(out / f"{table}.csv", index=False)
    print(table, df.shape)
con.close()
```
